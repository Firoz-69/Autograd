```python
words = open("names.txt" , 'r').read().splitlines()
```

```python
import math
import numpy as np 
import matplotlib.pyplot as plt
from graphviz import Digraph
```

```python
chars = sorted(list(set(''.join(words))))
stoi = {s:i+1 for i , s in enumerate(chars)}
stoi['.'] = 0
itos = {i:s for s , i in stoi.items()}
```

```python
import torch
N = torch.zeros((27,27) ,dtype=torch.int32)
N # here its an [28*28] matrix
```

```python

for w in words:
    chs = ['.'] + list(w) + ['.']
    for ch1 , ch2 in zip(chs ,chs[1:]):
        id1 , id2 = stoi[ch1] ,stoi[ch2]
        N[id1 ,id2] += 1 # here like [0,0] += 1

```

```python
probabilites = torch.zeros((27,27))
for i in range(27):
    totalSum = N[i].sum()

    for j in range(27):
        val = N[i][j]
        prob = val / totalSum
        probabilites[i][j] = prob
```

```python
ix = 1
print(probabilites[ix])
while True:
    p = probabilites[ix]
    # print(p)
    print(itos[ix], end='')

    ix = torch.multinomial(p, num_samples=1).item()
    # print(ix)

    if ix == 0:
        break

```
